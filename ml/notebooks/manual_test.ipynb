### Reading the result

- `vertical` means vertical separator lines and books counted left-to-right.
- `horizontal` means horizontal separator lines and books counted top-to-bottom.
- Increase **Min distance** if one physical edge is counted several times.
- Increase **Prominence** or **Peak height** to remove weak false boundaries.
- The confidence value is a provisional peak-quality score, not yet calibrated production confidence.
